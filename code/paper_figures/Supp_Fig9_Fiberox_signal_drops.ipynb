```python
import numpy as np
import matplotlib.pyplot as plt

# -----------------------------
# Global style: black background
# -----------------------------
plt.rcParams.update({
    "figure.facecolor": "black",
    "axes.facecolor": "black",
    "axes.edgecolor": "white",
    "axes.labelcolor": "white",
    "xtick.color": "white",
    "ytick.color": "white",
    "text.color": "white",
    "grid.color": "white",
})

# b-values (including b=0 reference)
bvals = np.array([0, 1000, 2000, 3000])

"""
Mean signals were manually extracted using MRtrix3:

dwiextract dwi.mif -bzero - | mrmath - mean b0.mif -axis 3 -force
dwiextract dwi.mif -shells 1000 - | mrmath - mean b1000.mif -axis 3 -force
dwiextract dwi.mif -shells 2000 - | mrmath - mean b2000.mif -axis 3 -force
dwiextract dwi.mif -shells 3000 - | mrmath - mean b3000.mif -axis 3 -force

White-matter masks:
- Phantoms: CSF-threshold–derived WM mask
- Human: brain mask

Signals are normalised to S0 (b=0).
"""

signals = {
    "HCP in vivo reference": [1.00, 0.48, 0.31, 0.23],
    "Phantom w/ fiber radius = 82.5µm": [1.00, 0.42, 0.29, 0.24],
    "Phantom w/ fiber radius= 52.7µm (automatic)": [1.00, 0.14, 0.08, 0.06],
    #"Phantom r=90.0": [1.00, 0.51, 0.38, 0.31],
}

plt.figure(figsize=(7, 7))

for label, vals in signals.items():
    if "HCP" in label:
        plt.plot(
            bvals,
            vals,
            linestyle="--",
            linewidth=3,
            color="white",
            marker="o",
            alpha=1.0,
            label=label,
            zorder=3
        )
    else:
        plt.plot(
            bvals,
            vals,
            linestyle="-",
            linewidth=8,
            marker="o",
            alpha=0.8,
            label=label,
            zorder=2
        )

plt.xlabel("b-value (s/mm²)", fontsize=16)
plt.ylabel("Normalised signal (S / S₀)", fontsize=16)
#plt.title("Signal drop as a function of b-value", fontsize=18)

ax = plt.gca()
ax.yaxis.tick_right()
ax.yaxis.set_label_position("right")
ax.tick_params(axis='y', which='both', left=False)
ax.spines['left'].set_visible(True)
ax.spines['right'].set_visible(True)
ax.spines['right'].set_color("white")

plt.ylim(0, 1.05)
plt.xlim(-0.05, 3100)
plt.xticks(bvals,fontsize=14)
plt.yticks(fontsize=14)


plt.grid(True, alpha=0.25)

legend = plt.legend(frameon=True, fontsize=14)
legend.get_frame().set_facecolor("black")
legend.get_frame().set_edgecolor("white")

plt.tight_layout()
plt.show()
```
